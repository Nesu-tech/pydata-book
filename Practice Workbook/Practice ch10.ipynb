**Scenario:** You have a DataFrame representing a small sales team.

```python
    'Sales': [1200, 950, 450, 500, 1150],
    'Quantity': [1, 1, 2, 1, 2]
})
```
**Task:** 1. Group the data by `City`.
2. Calculate the **mean** for the `Sales` column only.
3. **Goal:** Understand how to index a GroupBy object to avoid calculating means for irrelevant columns (like `Quantity`).

**Expected Solution:**
```python
# Students should identify that data.groupby('City')['Sales'].mean() 
# is more efficient than calculating the mean for the whole DF.

```


````python
import pandas as pd
import numpy as np

data = pd.DataFrame({
    'City': ['New York', 'London', 'London', 'New York', 'New York'],
    'Item': ['Laptop', 'Laptop', 'Tablet', 'Tablet', 'Laptop'],
    'Sales': [1200, 950, 450, 500, 1150],
    'Quantity': [1, 1, 2, 1, 2]

})
print(data)
grouped_df = data.groupby("City")["Sales"].mean()
grouped_df
````
