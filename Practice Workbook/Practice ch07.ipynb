### Problem 2: Pattern Hunting with Regex

**Scenario:** You are given a raw string containing a date in a specific format. You need to verify if it matches the pattern and extract the year.

**Task:** Using the `re` module, check if the string `data = "Report_Date: 2026-02-24"` contains a date in the format `YYYY-MM-DD`. Then, extract just the year (**2026**).

**Hints:**
